```python
import requests
import json
from config import API_KEY

BASE_URL = "http://api.openweathermap.org/data/2.5/weather"

def get_weather(city_name: str) -> dict:
    """
    دریافت اطلاعات آب و هوای یک شهر از OpenWeatherMap API
    
    Args:
        city_name: نام شهر به انگلیسی
        
    Returns:
        dict: دیکشنری حاوی اطلاعات آب و هوا
        
    Raises:
        Exception: اگر شهر پیدا نشود یا خطای API رخ دهد
    """
    params = {
        "appid": API_KEY,
        "q": city_name,
        "units": "metric"
    }
    
    try:
        response = requests.get(BASE_URL, params=params, timeout=10)
        response.raise_for_status()
        data = response.json()
        
        if data.get("cod") != 200:
            raise Exception(f"خطای API: {data.get('message', 'ناشناخته')}")
        
        weather = data["weather"][0]
        main = data["main"]
        
        return {
            "city": data["name"],
            "country": data["sys"]["country"],
            "temperature": main["temp"],
            "feels_like": main["feels_like"],
            "humidity": main["humidity"],
            "pressure": main["pressure"],
            "description": weather["description"],
            "wind_speed": data["wind"]["speed"],
        }
        
    except requests.exceptions.RequestException as e:
        raise Exception(f"خطای اتصال: {e}")
    except (KeyError, IndexError) as e:
        raise Exception("پاسخ API در قالب مورد انتظار نبود")

def main():
    city_name = input("نام شهر را وارد کنید (به انگلیسی): ").strip()
    
    if not city_name:
        print("نام شهر نمی‌تواند خالی باشد!")
        return
    
    try:
        weather = get_weather(city_name)
        print("\n" + "="*40)
        print(f"🌤  آب و هوای {weather['city']}, {weather['country']}")
        print("="*40)
        print(f"🌡  دما: {weather['temperature']}°C (احساس: {weather['feels_like']}°C)")
        print(f"💧 رطوبت: {weather['humidity']}%")
        print(f"🌀 فشار: {weather['pressure']} hPa")
        print(f"🌬  سرعت باد: {weather['wind_speed']} m/s")
        print(f"☁️  وضعیت: {weather['description']}")
    except Exception as e:
        print(f"❌ خطا: {e}")

if __name__ == "__main__":
    main()
```
